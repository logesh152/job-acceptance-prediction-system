```python
import pandas as pd
import numpy as np

df = pd.read_csv("Cleaned_Dataset.csv")
df.shape

df["experience_category"] = pd.cut(df["years_of_experience"],
                                   bins=[-1, 0, 2, 10],
                                   labels=["Fresher", "Junior", "Senior"]).astype(str)
print(df["experience_category"].value_counts())
print(df.groupby("experience_category")["status"].apply(lambda s: (s == "Placed").mean().round(3)))

df["academic_performance_band"] = pd.cut(
    df[["ssc_percentage", "hsc_percentage", "degree_percentage"]].mean(axis=1),
    bins=[0, 60, 75, 100], labels=["Low", "Medium", "High"]).astype(str)
print(df["academic_performance_band"].value_counts())

df["skills_match_level"] = pd.cut(df["skills_match_percentage"],
                                  bins=[0, 60, 85, 100],
                                  labels=["Low", "Medium", "High"]).astype(str)
print(df["skills_match_level"].value_counts())
print(df.groupby("skills_match_level")["status"].apply(lambda s: (s == "Placed").mean().round(3)))

df["interview_performance_category"] = pd.cut(
    df[["technical_score", "aptitude_score", "communication_score"]].mean(axis=1),
    bins=[0, 55, 70, 100], labels=["Weak", "Average", "Strong"]).astype(str)
print(df["interview_performance_category"].value_counts())
print(df.groupby("interview_performance_category")["status"].apply(lambda s: (s == "Placed").mean().round(3)))

df["ctc_expectation_gap_lpa"] = df["expected_ctc_lpa"] - df["previous_ctc_lpa"]
df[["previous_ctc_lpa", "expected_ctc_lpa", "ctc_expectation_gap_lpa"]].head()

df.to_csv("Final_Dataset.csv", index=False)
print("Saved Final_Dataset.csv :", df.shape)
```
