```python
import pandas as pd
import numpy as np

df = pd.read_csv("Dataset.csv")
print("Dataset shape:", df.shape)
df.head()

df.info()

df.isnull().sum()

df.describe().T

print("Duplicate rows:", df.duplicated().sum())
df = df.drop_duplicates()
print("After removing duplicates:", df.shape)

cat_cols = df.select_dtypes(include="object").columns
for col in cat_cols:
    print(col, "->", df[col].unique())

# gender: small letters -> proper case
df["gender"] = df["gender"].replace({"male": "Male", "female": "Female"})

# internship: remove spaces and capitalise
df["internship_experience"] = df["internship_experience"].str.strip().str.title()

# strip trailing spaces from tier columns
for col in ["company_tier", "job_role_match", "competition_level"]:
    df[col] = df[col].str.strip()

# check again
for col in ["gender", "internship_experience", "company_tier"]:
    print(col, "->", df[col].unique())

print("New duplicate rows after standardizing:", df.duplicated().sum())
df = df.drop_duplicates()
print("Final rows:", df.shape)

df["ssc_percentage"] = df["ssc_percentage"].fillna(df["ssc_percentage"].median())
df["hsc_percentage"] = df["hsc_percentage"].fillna(df["hsc_percentage"].median())
df["notice_period_days"] = df["notice_period_days"].fillna(df["notice_period_days"].median())

df["employment_gap_months"] = df["employment_gap_months"].fillna(0)

for col in ["career_switch_willingness", "relevant_experience", "job_role_match",
            "layoff_history", "relocation_willingness"]:
    df[col] = df[col].fillna(df[col].mode()[0])

df.isnull().sum().sum()   # total missing values left - should be 0

freshers = df[df["years_of_experience"] == 0]
print("Freshers with previous CTC:", (freshers["previous_ctc_lpa"] > 0).sum())

df.loc[df["years_of_experience"] == 0, "previous_ctc_lpa"] = 0

df.to_csv("Cleaned_Dataset.csv", index=False)
print("Saved Cleaned_Dataset.csv :", df.shape)
print("Placement rate:", round((df["status"] == "Placed").mean() * 100, 2), "%")
```
